```python
ptf = get_positions("PF72354", "01/01/2025")
```

```python
ptf["positionWeight"] = ptf["positionWeight"].astype(float)
test = ptf[["positionWeight", "instIssuerCountry"]]
test.groupby(by="instIssuerCountry").sum().sort_values(
    by="positionWeight", ascending=False
).round(4)
```

```python
ptf["positionWeight"] = ptf["positionWeight"].astype(float)
test = ptf[["positionWeight", "instSectorBicsName1"]]
test.groupby(by="instSectorBicsName1").sum().sort_values(
    by="positionWeight", ascending=False
).round(4)
```

```python
bins = [
    0,
    2000000000,
    10000000000,
    200000000000,
    float("inf"),
]  # Added float('inf') for Mega Cap
labels = [
    "Small Cap (200m - 2b)",
    "Mid Cap (2b - 10b)",
    "Large Cap (10b - 200b)",
    "Mega Cap (> $ 200b)",
]

# Create a copy of the relevant columns
test = ptf[["positionWeight", "instMktCap"]].copy()
test["instMktCap"] = test["instMktCap"].astype(float)

# Map the instMktCap to the defined labels
test["Style"] = pd.cut(test["instMktCap"], bins=bins, labels=labels, right=False)
test[["positionWeight", "Style"]].groupby(by="Style").sum().round(4)
```
